# 2025 Descriptive statistics


```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import pandas as pd
```

```python
data = pd.read_csv("../data/sam3/cleaned_data_anon_teams.csv", low_memory=False)
```

```python
# Limit to 2025 data
data = data[data['year'] == 2025]

# Add columns
# Add columns
data['nihss>10'] = (data['stroke_severity'] > 10).astype(int)
data['nihss>10_or_lvo'] = (
    (data['stroke_severity'] > 10) | (data['lvo'] == 1)
).astype(int)
data['discharge_mrs_0-1'] = (
    (data['discharge_disability'] >= 0) &
    (data['discharge_disability'] <= 1)
).astype(int)
data['discharge_mrs_0-2'] = (
    (data['discharge_disability'] >= 0) &
    (data['discharge_disability'] <= 2)
).astype(int)
```

```python
descriptive_stats_all_2025 = data.describe().T
descriptive_stats_all_2025
```

```python
# Divide the data into two groups based on the 'thrombectomy_centre' column, and produce descriptive statistics for each group
thrombectomy_centre_data = data[data['thrombectomy_centre'] == 1]
non_thrombectomy_centre_data = data[data['thrombectomy_centre'] == 0]

descriptive_stats_thrombectomy_centre = thrombectomy_centre_data.describe().T
descriptive_stats_non_thrombectomy_centre = non_thrombectomy_centre_data.describe().T
```

```python
descriptive_stats_thrombectomy_centre
```

```python
descriptive_stats_non_thrombectomy_centre
```
