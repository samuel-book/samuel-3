# Graphviz plot of Directed Acyclic Graph (DAG) for thrombolysis use in stroke

An example DAG showing key determinants for outcome in stroke. This includes use of thrombolysis (a *clot-busting* treatment), and the patient characteristics that influence whether thrombolysis is used.

## Set up graph

```python
from graphviz import Digraph
```

## Thrombolysis only (mild stroke)

```python
# Create a directed graph
g = Digraph()

# Set graph attributes
g.attr(overlap='scale')
g.attr(sep='0.25')
g.attr(layout='dot') # See https://graphviz.org/docs/layouts/

# Set node attributes
g.attr('node', shape='circle', fixedsize='true', width='1.4', height='1.4')
```

```python
g.node('outcome', label='Outcome', penwidth='2', color='red', fontname='times bold', fontcolor='Red')
g.node('nihss', label='Stroke\nseverity', color='black', fontcolor='Black')
g.node('diabetes', label='Diabetes', color='black', fontcolor='Black')
g.node('hypertension', label='Hypertension', color='black', fontcolor='Black')
g.node('congestive_heart_failure', label='Congestive\nheart failure', color='black', fontcolor='Black')
g.node('team', label='Stroke\nteam', color='black', fontcolor='Black')
#g.node('scan', label='Arrival to\nscan time', color='black', fontcolor='Black')
g.node('mrs', label='Prior\ndisability', color='black', fontcolor='Black')
g.node('age', label='Age', color='black', fontcolor='Black')
#g.node('known', label='Known onset\ntime', color='black', fontcolor='Black')
g.node('af', label='Atrial\nfibrillation', color='black', fontcolor='Black')
g.node('anticoag', label='Anticoagulation\nuse', color='black', fontcolor='Black')
g.node('thrombolysis', label='Thrombolysis\n(use/time)', penwidth='2', color='red', fontname='times bold', fontcolor='Red')
#g.node('thrombectomy', label='Thrombectomy\n(use/time)', penwidth='2', color='red', fontname='times bold', fontcolor='Red')
#g.node('precise', label='Estimated or \nprecicely known\nonset time', color='black', fontcolor='Black')
```

```python
g.edge('nihss', 'outcome', color='Black') # style='dashed')
g.edge('mrs', 'outcome', color='Black')
g.edge('age', 'outcome', color='Black')
g.edge('af', 'outcome', color='Black', style='dashed')
g.edge('diabetes', 'outcome', color='Black', style='dashed')
g.edge('hypertension', 'outcome', color='Black', style='dashed')
g.edge('thrombolysis', 'outcome', color='Red')
g.edge('team', 'outcome', color='Black')
g.edge('congestive_heart_failure', 'outcome', color='Black', style='dashed')

# Show confounders
g.edge('nihss', 'thrombolysis', color='Black')
g.edge('mrs', 'thrombolysis', color='Black')
g.edge('age', 'thrombolysis', color='Black', style='dashed')
g.edge('hypertension', 'thrombolysis', color='Black', style='dashed')
g.edge('anticoag', 'thrombolysis', color='Black')
g.edge('team', 'thrombolysis', color='Black')
g.edge('congestive_heart_failure', 'thrombolysis', color='Black', style='dashed')
g.edge('diabetes', 'thrombolysis', color='Black', style='dashed')

# Others
g.edge('age', 'mrs', color='Black')
g.edge('age', 'nihss', color='Black')
g.edge('age', 'af', color='Black')
g.edge('age', 'hypertension', color='Black')
g.edge('age', 'congestive_heart_failure', color='Black')
g.edge('age', 'diabetes', color='Black')
g.edge('af', 'anticoag', color='Black')
```

```python
g
```

```python
# Save the graph as a PNG file
g.format = 'png'
g.render('thrombolysis_mild_stroke_early_arrival_dag', cleanup=True)
```

## Thrombolyssis + Thrombectomy (non-mild stroke)

```python
# Create a directed graph
g = Digraph()

# Set graph attributes
g.attr(overlap='scale')
g.attr(sep='0.25')
g.attr(layout='dot') # See https://graphviz.org/docs/layouts/

# Set node attributes
g.attr('node', shape='circle', fixedsize='true', width='1.4', height='1.4')
```

```python
g.node('outcome', label='Outcome', penwidth='2', color='red', fontname='times bold', fontcolor='Red')
g.node('nihss', label='Stroke\nseverity', color='black', fontcolor='Black')
g.node('diabetes', label='Diabetes', color='black', fontcolor='Black')
g.node('hypertension', label='Hypertension', color='black', fontcolor='Black')
g.node('congestive_heart_failure', label='Congestive\nheart failure', color='black', fontcolor='Black')
g.node('team', label='Stroke\nteam', color='black', fontcolor='Black')
#g.node('scan', label='Arrival to\nscan time', color='black', fontcolor='Black')
g.node('mrs', label='Prior\ndisability', color='black', fontcolor='Black')
g.node('age', label='Age', color='black', fontcolor='Black')
#g.node('known', label='Known onset\ntime', color='black', fontcolor='Black')
g.node('af', label='Atrial\nfibrillation', color='black', fontcolor='Black')
g.node('anticoag', label='Anticoagulation\nuse', color='black', fontcolor='Black')
g.node('thrombolysis', label='Thrombolysis\n(use/time)', penwidth='2', color='red', fontname='times bold', fontcolor='Red')
g.node('thrombectomy', label='Thrombectomy\n(use/time)', penwidth='2', color='red', fontname='times bold', fontcolor='Red')
#g.node('precise', label='Estimated or \nprecicely known\nonset time', color='black', fontcolor='Black')
```

```python
g.edge('nihss', 'outcome', color='Black') # style='dashed')
g.edge('mrs', 'outcome', color='Black')
g.edge('age', 'outcome', color='Black')
g.edge('af', 'outcome', color='Black', style='dashed')
g.edge('diabetes', 'outcome', color='Black', style='dashed')
g.edge('hypertension', 'outcome', color='Black', style='dashed')
g.edge('thrombolysis', 'outcome', color='Red')
g.edge('thrombectomy', 'outcome', color='Red')
g.edge('team', 'outcome', color='Black')
g.edge('congestive_heart_failure', 'outcome', color='Black', style='dashed')

# Show confounders
g.edge('nihss', 'thrombolysis', color='Black')
g.edge('mrs', 'thrombolysis', color='Black')
g.edge('age', 'thrombolysis', color='Black', style='dashed')
g.edge('hypertension', 'thrombolysis', color='Black', style='dashed')
g.edge('anticoag', 'thrombolysis', color='Black')
g.edge('team', 'thrombolysis', color='Black')
g.edge('congestive_heart_failure', 'thrombolysis', color='Black', style='dashed')
g.edge('diabetes', 'thrombolysis', color='Black', style='dashed')

g.edge('nihss', 'thrombectomy', color='Black')
g.edge('mrs', 'thrombectomy', color='Black')
g.edge('age', 'thrombectomy', color='Black', style='dashed')
g.edge('hypertension', 'thrombectomy', color='Black', style='dashed')
g.edge('team', 'thrombectomy', color='Black')
g.edge('thrombolysis', 'thrombectomy', color='Black')
g.edge('diabetes', 'thrombolysis', color='Black', style='dashed')


# Others
g.edge('age', 'mrs', color='Black')
g.edge('age', 'nihss', color='Black')
g.edge('age', 'af', color='Black')
g.edge('age', 'hypertension', color='Black')
g.edge('age', 'congestive_heart_failure', color='Black')
g.edge('age', 'diabetes', color='Black')
g.edge('af', 'anticoag', color='Black')

# Put thrombolysis and thrombectomy at same level
with g.subgraph() as s:
    s.attr(rank='same')
    s.node('thrombolysis')
    s.node('thrombectomy')
```
